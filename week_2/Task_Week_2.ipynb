**HyperParameter Tuning:**

```python
param_grid = {'max_depth': [2, 3, 4, 5, 6, 7, 8, None]}
```

```python
from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid, cv=5)
```

```python
grid_search.fit(x_train, y_train)
```

```python
print("Best max_depth:", grid_search.best_params_)
print("Best cross-validation accuracy:", grid_search.best_score_)
```

**Tuning DecisionTree overfit:**

```python
best_dt_model = DecisionTreeClassifier(max_depth=3, random_state=42)
best_dt_model.fit(x_train, y_train)
print("Train accuracy:", best_dt_model.score(x_train, y_train))
print("Test accuracy:", best_dt_model.score(x_test, y_test))
```
