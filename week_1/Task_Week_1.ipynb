```python
df["Transmission"].fillna(df["Transmission"].mode()[0], inplace=True)
```

```python
df["Accident_History"] = df["Accident_History"].fillna(df["Accident_History"].mode()[0])
```

```python
df["Service_History"]=df["Service_History"].fillna(df["Service_History"].mode()[0])
```

```python
df["Color"] = df["Color"].fillna(df["Color"].mode()[0])
```

```python
df["Location"] = df["Location"].fillna(df["Location"].mode()[0])
```

```python
df["Engine_Size"] = df["Engine_Size"].fillna(df["Engine_Size"].median())
```

```python
df["Horsepower"] = df["Horsepower"].fillna(df["Horsepower"].median())
```

```python
df["Torque"] = df["Torque"].fillna(df["Torque"].median())
```

```python
df["Fuel_Efficiency"] = df["Fuel_Efficiency"].fillna(df["Fuel_Efficiency"].median())
```
