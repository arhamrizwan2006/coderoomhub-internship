## Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaled_features = scaler.fit_transform(df)
```

## Dimensionality Reduction with PCA

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca_features = pca.fit_transform(scaled_features)
```

```python
print(pca.explained_variance_ratio_)
```

```python
print(sum(pca.explained_variance_ratio_))
```

## K-Means Clustering using Elbow Method

```python
from sklearn.cluster import KMeans
inertia_values = [ ]

for k in range(1,11):
    kmeans=KMeans(n_clusters=k, random_state=42,n_init=10)
    kmeans.fit(scaled_features)
    inertia_values.append(kmeans.inertia_)
```

```python
plt.plot(range(1,11),inertia_values, marker='o')
plt.xlabel('Number of Clusters(k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
kmeans_final =KMeans(n_clusters=5, random_state=42, n_init=10)
```

```python
kmeans_labels = kmeans_final.fit_predict(scaled_features)
```

```python
df['KMeans_Cluster'] = kmeans_labels
```

```python
df.info()
```

```python
df['KMeans_Cluster'].head()
```

```python
df['KMeans_Cluster'].unique()
```

```python
df['KMeans_Cluster'].value_counts()
```

## Visualizing K-Means Clusters

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_features[:,0],pca_features[:,1], c=kmeans_labels, cmap='viridis')
plt.xlabel('Principal Compenent 1')
plt.ylabel('Principal Compenent 2')
plt.title('Customer Segments (K-Means, k=5) — PCA Visualization')
plt.colorbar(label='Cluster')
plt.show()
```

## Hierarchical Clustering (Dendrogram)

```python
from scipy.cluster.hierarchy import dendrogram, linkage
linked = linkage(scaled_features, method='ward')

plt.figure(figsize=(10,6))
dendrogram(linked)
plt.title('Hierarchical Clustering Dendogram')
plt.xlabel('Customers')
plt.ylabel('Distance')
plt.show()
```

## Fitting Agglomerative Clustering

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
hierarchical = AgglomerativeClustering(n_clusters=5, linkage ='ward')
hierarchical_labels = hierarchical.fit_predict(scaled_features)
```

```python
df['Hierarchical_Cluster'] = hierarchical_labels
```

```python
df.info()
```

```python
df['Hierarchical_Cluster'].value_counts()
```

## Comparison b/w K-Means & Hierarchical Clustering

```python
comparison = pd.crosstab (df['KMeans_Cluster'],df['Hierarchical_Cluster'])
```

```python
print(comparison)
```

## Visualizing Hierarchical Clusters

```python
plt.figure(figsize=(8,6))
plt.scatter(pca_features[:,0], pca_features[:,1], c=hierarchical_labels, cmap='plasma')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Customer Segments (Hierarchical, k=5) — PCA Visualization')
plt.colorbar(label='Cluster')
plt.show()
```

## DBSCAN Choosing eps 

```python
from sklearn.neighbors import NearestNeighbors
neighbors = NearestNeighbors(n_neighbors=5)
neighbors_fit= neighbors.fit(scaled_features)
distances, indices = neighbors_fit.kneighbors(scaled_features)
```

```python
distances=np.sort(distances[:,4])
```

## Visualizing k-distance graph

```python
plt.figure(figsize=(8, 6))
plt.plot(distances)
plt.xlabel('Points sorted by distance')
plt.ylabel('5th Nearest Neighbor Distance')
plt.title('K-Distance Graph for DBSCAN eps Selection')
plt.show()
```

## DBSCAN Clustering & Anomaly Detection

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.9, min_samples=5)
dbscan_labels = dbscan.fit_predict(scaled_features)

df['DBSCAN_Cluster'] = dbscan_labels

print(pd.Series(dbscan_labels).value_counts())
```

## Visualizing DBSCAN Clusters and Noise Points

```python
plt.figure(figsize=(8, 6))
plt.scatter(pca_features[:, 0], pca_features[:, 1], c=dbscan_labels, cmap='plasma')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('DBSCAN Clustering with Noise Points (Anomalies)')
plt.colorbar(label='Cluster (-1 = Noise/Anomaly)')
plt.show()
```

## Cluster Profiling by Interpreting the Segments

```python
cluster_profile = df.groupby('KMeans_Cluster')[['Age', 'Annual Income (k$)', 'Spending Score (1-100)']].mean()
print(cluster_profile)
```

```python
gender_breakdown = df.groupby('KMeans_Cluster')['Gender'].mean()
print(gender_breakdown)
```
